```python
import jax.numpy as jnp


i = jnp.array([0, 0, 1, 0, 1, 3])
a = jnp.array(range(len(i)))
perm = jnp.argsort(i)
i = i[perm]
print(i)
counts = jnp.bincount(i)#.tolist()
print(counts)
cum = jnp.cumsum(counts)
print(cum)
jnp.split(a[perm], cum[:-1])
```
