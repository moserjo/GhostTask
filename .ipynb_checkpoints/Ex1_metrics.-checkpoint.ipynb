# Error metrics provisorisch Ex1
zero tasks: 10x10 grid (da ja exakt bekannt)

```python
runs = 10

sigma_list = [0.0, 0.01, 0.1, 0.2]
n_list = [3, 5, 10]#,  30, 50]
modes = ["GT", "PIGP"]
all_AE = np.zeros((len(n_list), len(sigma_list), len(modes), 1, runs))



for n_idx, npoints in enumerate(n_list):
    for sigma_idx, sigma in enumerate(sigma_list):
        for j, mode in enumerate(modes):
            for run in range(runs):
                if npoints == 30:
                    data_path_ex1 = os.path.join(output_data_ex1, "Ex1_n%i_sigma%.2f_%s_%s.npz"%(3, sigma, mode, "inverse"))
                elif npoints == 50:
                    data_path_ex1 = os.path.join(output_data_ex1, "Ex1_n%i_sigma%.2f_%s_%s.npz"%(5, sigma, mode, "inverse"))
                else:
                    data_path_ex1 = os.path.join(output_data_ex1, "Ex1_n%i_sigma%.2f_%s_%s_%i.npz"%(npoints, sigma, mode, "inverse_extra", run))
                data_ex1 = np.load(data_path_ex1)
                parameters_during_training = data_ex1["parameters_during_training"]
                used_parameters = data_ex1["parameter_order"]
                for i, key in enumerate(used_parameters):
                    if key == "a":
                        learned_parameters = parameters_during_training[i, -1]
                
                all_AE[n_idx, sigma_idx, j, 0, run] = abs(learned_parameters-2.) 
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --- reshape all_AE into long-form dataframe ---
records = []

display_n = ["3", "5", "10", "3noex", "5noex"]

for n_idx, n in enumerate(n_list):
    for sigma_idx, sigma in enumerate(sigma_list):
        for mode_idx, mode in enumerate(modes):
            for run in range(runs):
                ae = all_AE[n_idx, sigma_idx, mode_idx, 0, run]
                records.append({
                    "n": display_n[n_idx],
                    "sigma": sigma,
                    "mode": mode,
                    "AE": ae
                })

df = pd.DataFrame(records)

# --- plotting ---
fig, axes = plt.subplots(1, 4, figsize=(22, 6))#, sharey=True)
axes = axes.flatten()

for ax, sigma in zip(axes, sigma_list):
    subset = df[df["sigma"] == sigma]

    sns.boxplot(
        data=subset,
        x="n",
        y="AE",
        hue="mode",
        ax=ax
    )

    ax.set_title(f"σ = {sigma:.2f}")
    ax.set_yscale("log")
    ax.set_xlabel("n")
    ax.set_ylabel("Absolute Error")

# cleaner legend: keep only first subplot legend
for ax in axes[1:]:
    ax.get_legend().remove()

axes[0].legend(title="Mode")

plt.suptitle("Absolute MAP Errors", fontsize=16)
plt.tight_layout()
plt.show()
```
